```python
## Day 2- Nutrition Dataset Exploration
```

```python
import pandas as pd

df= pd.read_csv("/content/drive/MyDrive/Colab Notebooks/Indian_Food_Nutrition_Processed.csv")
print(df.shape)
print(df.head())
```

```python
df.info()
df.describe()
df.isnull().sum()
```

```python
df.columns= df.columns.str.strip()
df= df.drop_duplicates(subset="Dish Name")
df=df.dropna(subset=['Calories (kcal)'])
print(df.shape)
```

```python
def search_food(name, df):
  result= df[df['Dish Name'].str.contains(name, case=False, na= False)]
  if result.empty:
    return f"No match for '{name}'"
  return result[['Dish Name', 'Calories (kcal)', 'Protein (g)', 'Carbohydrates (g)', 'Fats (g)']]

search_food("paneer", df)
```

```python
import matplotlib.pyplot as plt

top10= df.nlargest(10, 'Calories (kcal)')
top10.plot(x='Dish Name', y='Calories (kcal)', kind='barh', legend= False)
plt.xlabel('Calories (kcal)')
plt.title('Top 10 Highest Calorie Indian Dishes (per 100g)')
plt.tight_layout()
plt.show()
```

Data was clean — no duplicate dishes, no missing values in core nutrition columns (only Vitamin C and Folate had some gaps, which weren't dropped since they're not the main focus).
Built a search function to look up any dish and see its nutrition profile.
Next question to explore: how does this data compare to what a generic nutrition tool (like MyFitnessPal) estimates for the same dish?

```python
comparison_data= {
    'Dish Name': ['Dal Makhani', 'Bhatura', 'Shahi Paneer'],
    'Dataset_Calories': [74.04, 793.2, 156.5],
    'Google_Calories': [140, 335, 155 ]
}
comparison_df= pd.DataFrame(comparison_data)
print(comparison_df)
```

```python
import matplotlib.pyplot as plt

comparison_df.plot(x='Dish Name', y=['Dataset_Calories', 'Google_Calories'], kind= 'bar')
plt.ylabel('Calories(kcal)')
plt.title('Dataset vs Google: Calorie Estimates')
plt.tight_layout()
plt.show()
```

## Day 3 Findings
Compared 3 dishes between the dataset and Google's calorie estimates, both normalized
to per-100g values to remove portion-size bias.

Dal Makhani and Shahi Paneer showed reasonably close values across both sources.
Bhatura, however, showed a large and persistent gap — the dataset reports ~800 kcal
per 100g versus Google's ~330 kcal per 100g, even after controlling for serving size.

This suggests genuine inconsistency in how different sources estimate calories for
fried, oil-heavy dishes like Bhatura — possibly due to differing assumptions about oil
absorption during frying, which varies a lot in practice and is hard to standardize.
This is a promising direction to investigate further with more fried dishes in the dataset.
